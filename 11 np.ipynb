```python
import numpy as np

def generate_otp(length=4):
    otp = ''.join([str(np.random.randint(0, 10)) for _ in range(length)])
    return otp

print("Your OTP is:", generate_otp())
```

```python
import random

def guessing_game():
    number = random.randint(1, 10)

    print("I selected a number between 1 and 10.")

    guess = int(input("Enter your guess: "))

    if guess == number:
        print("🎉 Correct!")
    else:
        print("❌ Wrong!")
        print("The number was:", number)

guessing_game()
```
