
## copy list

```python
list2
```

```python
list8 = list2 # Create a new reference "list8"
```

```python

id(list8) , id(list2) # The address of both mylist & mylist1 will be the same
```

```python

list9 = list2.copy() # Create a copy of the list
```

```python
id(list9) , id(list2)
```

```python
list8[0] = 55  # mylist1 will be also impacted as it is pointing to the same list
print(list8,list2)
```

```python

list9[0] = 80  # Copy of list won't be impacted due to changes made on the original lis
print(list9,list2)
```
