
# Left Shift Operator <<
## The left shift operator (<<) moves all the bits to the left by a specified number of positions.

## Example
   #### 5 << 1
   #### 5 = 1 0 1 
   #### when we do left shift with  1 then it becomes 

 #### 5 = 1 0 1 0

 #### now ans  = 10

```python
5<<1
```

```python
10<<1
```

```python
222<<333
```


# Right Shift Operator >>
### The right shift operator (>>) moves all the bits to the right by a specified number of positions.

## Example

```python
5>>1
```

```python
21>>1
```

```python
11>>2
```
