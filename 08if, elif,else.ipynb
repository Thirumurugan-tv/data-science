### 1. if

##### if executes a block of code when the condition is True

```python
age = 20

if age >=18:
    print('Eligible')
```

### 2. Multple if Statements

##### When we use multiple if statements, each condition is checked independently.

```python
age = 20

if age >=18:
    print('Eligible')
else:
    print("Not eligible")
```

## 3. if-else

##### else is used when there are two possible outcomes.

```python
age = 20
if age >=18:
    print("Eligible")
else:
    print("Not eligible")
```

## 4. if-elif-else
##### elif means "else if".

##### It is used when we have more than two possible conditions.

##### Use if-elif-else when you have multiple mutually exclusive choices.

##### Because Python checks the conditions one by one and stops when it finds the first True condition.

```python
marks =75

if marks >=90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >=60:
    print("C")
else:
    print("Fail")
```
